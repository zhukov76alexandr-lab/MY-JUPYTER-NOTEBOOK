Переменные

КАК НАЗЫВАТЬ ПЕРЕМЕННЫЕ
1) Имя переменной обычно называют буквами, можно использовать числа и подчёркивание
2) Имя должно описывать суть (на английском языке)
3) не может начинаться с цифры
4) не может использовать заглавные буквы (snake_case).
   Иногда применяются заглавные без подчёркивания: CamelCase (в Python бычно для обозначения классов).
