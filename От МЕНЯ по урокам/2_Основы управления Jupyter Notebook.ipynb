Про выполнение команд
* Shift + Enter - выполнить код в ячейке и перейти на следующую
* Ctrl + Enter - выполнить и остаться

Немного полезностей и по markdown:
* для автозаплнения длинных названий используйте табуляцию
* перевести ячейку из кода в комментарии 'Esc + m' или просто 'm', если вы ещё не успели сделать её активной
* перевести ячейку из комментария в код: 'Esc + y'
* добавить новую ячейку можно аналогичным образом через 'b'
  Jupyter понимает эти команды и в русской раскладке
  всё это можно сделать и через верхнюю панель под меню File, Elit, ...
* список горячих клавиш Ctrl+Shift+C (Command+Shift+P для мака)
