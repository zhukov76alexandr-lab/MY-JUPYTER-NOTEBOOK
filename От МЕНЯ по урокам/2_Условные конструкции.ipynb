Операторы сравнения

;   > - больше
;   < - меньше
;   == - проверка на тождество
;   >= - больше или равно
;   <= - меньше или равно
;   != - неравенство   (в др. языках может быть: <>)
;   = - равенство

Проверка чётности числа

```python
### проверка четности числа
x = 6
if x % 2 == 0:
    print(x, '-чётное число')
```

```python
### проверка чётности числа с условием невыполнения
x = 5
if x % 2 == 0:
    print(x, '- чётное число')
else:
    print(x, '- нечётное число')
```

```python
### AND (True - оба условия верны)
1 < 2 and 2 < 3
```

```python
0 < 2 and 3 < 2
```

```python
### OR (True - или одно верно, или другое)
0 < 2 or 3 < 2
```

```python
0 > 1 or 1 > 2
```

```python
### NOT ("НЕ" когда нужно обратить результат проверки в обратную сторону)
0 < 1
```
